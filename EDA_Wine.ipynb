```python
df_quality_FA = df.groupby("quality")["fixed acidity"].mean().sort_values(ascending=False)
```

```python
df_quality_alcohol = df.groupby("quality")["alcohol"].mean().sort_values(ascending=False)
```

```python
df_quality_density = df.groupby("quality")["density"].mean().sort_values(ascending=False)
```
