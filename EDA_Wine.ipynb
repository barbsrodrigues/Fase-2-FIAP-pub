```python
import pandas as pd
df = pd.read_csv("WineQT.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df_quality_FA = df.groupby("quality")["fixed acidity"].mean().sort_values(ascending=False)
```

```python
df_quality_alcohol = df.groupby("quality")["alcohol"].mean().sort_values(ascending=False)
```

```python
df_quality_density = df.groupby("quality")["density"].mean().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
```

```python
df_quality_density
```

```python
df_quality_alcohol
```

```python
df_quality_FA
```

```python
df_quality_FA.plot(kind="bar")
plt.xlabel("Fixed Acidity")
plt.ylabel("Quality")
plt.title("Fixed Acidity vs Quality")
```

```python
n_quality = df['quality'].value_counts()
plt.barh(n_quality.index, n_quality.values)
plt.xlabel('Count')
plt.ylabel('Quality')
plt.title('Count of Quality')
```

```python
import seaborn as sns
```

```python
sns.set(style="whitegrid")
fig, axes = plt.subplots(figsize=(15, 5))
sns.boxplot(x="alcohol", data=df)
axes.set_title("Alcohol")
plt.show()
```

```python
sns.set(style="whitegrid")
fig, axes = plt.subplots(figsize=(15, 5))
sns.boxplot(x="pH", data=df)
axes.set_title("pH")
plt.show()
```

```python
sns.set(style="whitegrid")
fig, axes = plt.subplots(figsize=(15, 5))
sns.boxplot(x="fixed acidity", data=df)
axes.set_title("F.A")
plt.show()
```
